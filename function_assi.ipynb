##Q5.What are iterators in python and how do they differ from iterables?

Ans.
   ###  **Theoretical Explanation: Iterators vs Iterables in Python**

In Python, **iterators** and **iterables** are both used to loop through data, but they represent different concepts. Understanding the difference is essential for working with loops, generators, and memory-efficient programming.

---

##  **What is an Iterable?**

An **iterable** is any Python object capable of returning its elements one at a time. It is an object that can be **looped over** using a `for` loop or by converting it into an iterator using the built-in `iter()` function.

###  Characteristics of an Iterable:
- Implements the `__iter__()` method.
- Returns an **iterator** when `iter()` is called on it.
- Examples: `list`, `tuple`, `string`, `set`, `dict`.

---

## **What is an Iterator?**

An **iterator** is an object that **represents a stream of data**, and it **remembers its position** during iteration. It is created from an iterable using the `iter()` function.

###  Characteristics of an Iterator:
- Implements **two methods**:
  - `__iter__()` → returns the iterator itself.
  - `__next__()` → returns the next value from the iterable.
- Raises `StopIteration` when there are no more elements to return.
- Can be used **only once**, since it does not reset automatically.

---

##  **Key Differences:**

| Feature               | Iterable                            | Iterator                            |
|-----------------------|--------------------------------------|--------------------------------------|
| Purpose               | Can return an iterator               | Used to get the next value           |
| Key Methods           | `__iter__()`                         | `__iter__()` and `__next__()`        |
| Memory                | Does not track state                 | Remembers current state              |
| Examples              | `list`, `str`, `tuple`, etc.         | Object returned by `iter()`          |
| Usage                 | Can be used multiple times in loops  | Used once unless re-created          |

---

### Summary:

- An **iterable** is like a book (can be read multiple times).
- An **iterator** is like a bookmark (remembers where you left off).

##Q6.Explain the concept of generators in python and how they are defined.

Ans.
  ###  **Theoretical Explanation: Generators in Python**

A **generator** in Python is a special type of **iterator** that is used to generate values **one at a time**, **on demand**, rather than computing and storing them all at once in memory. This makes generators **highly memory-efficient**, especially when working with large datasets or infinite sequences.

---

### **Definition and Key Concepts:**

1. **Generators are defined using a normal function with the `yield` keyword** instead of `return`.
2. When a generator function is called, it **does not run the code immediately**; instead, it returns a **generator object**.
3. Each time the `yield` statement is executed, the function’s state is **paused** and saved.
4. When `next()` is called, execution **resumes from where it left off**, continues until the next `yield`, and **returns the next value**.
5. Once all values are exhausted, a `StopIteration` exception is raised.

---

### **Important Characteristics:**

- **Lazy Evaluation**: Values are generated one at a time, only when requested.
- **State Preservation**: The generator remembers its place between calls.
- **No Need for a Class**: You can create iterators without writing a class and implementing `__iter__()` and `__next__()`.

---

### **Why Use Generators?**

- **Efficiency**: They use less memory because values are generated on the fly.
- **Performance**: Faster for iterating through large or infinite sequences.
- **Convenience**: Easier to write and maintain compared to custom iterator classes.

---

###  Summary:

| Feature                   | Description |
|---------------------------|-------------|
| Defined with              | `def` + `yield` |
| Returns                   | A generator object |
| Used with                 | `for` loops or `next()` |
| Saves function state?     | Yes |
| Memory usage              | Low (lazy evaluation) |
| Stops with                | `StopIteration` |

---


##Q7.What are the advantage of using generators over regular functions.

Ans.
   ###  **Theoretical Advantages of Using Generators Over Regular Functions in Python**

Generators offer distinct advantages over regular functions, particularly when dealing with large datasets, streaming data, or scenarios requiring high memory efficiency. Here's a detailed theoretical explanation of these benefits:

---

##  **1. Memory Efficiency**

- **Generators do not store the entire sequence in memory**.
- They yield values **one at a time** as they are requested, which means that **only one value** is in memory at any given time.
- **Regular functions** (like those returning lists) must **compute and store all values** before returning, which can lead to high memory usage for large datasets.

### Example:
- If you have a large dataset or are processing a file line by line, using a generator allows you to process the data **without loading it all into memory** at once.

---

##  **2. Lazy Evaluation**

- **Lazy evaluation** means that values are **not computed until they are needed**.
- Generators are **lazy**, meaning they produce items **only when requested**, which is a major advantage when not all of the data needs to be processed upfront.
- This behavior can **improve performance** by preventing unnecessary calculations or I/O operations.

### Example:
- If you only need the first few results of a large sequence, a generator will **stop processing as soon as you get the required values**, instead of computing the entire sequence.

---

##  **3. Preservation of State**

- **Generators automatically preserve their state** between calls to `next()`.
- When a generator's `yield` is encountered, the function **pauses** and saves its current state (variable values, execution point), so that it can **resume** later exactly where it left off.
- This eliminates the need for manual state management, as is necessary in **regular functions** or **iterator classes**.

### Example:
- A generator doesn’t need any additional logic to remember where it was in a sequence—it automatically picks up from where it left off each time you call `next()`.

---

##  **4. Simplified Code**

- Using a generator eliminates the need to manually write complex iteration logic or use auxiliary data structures to store the intermediate values.
- **Regular functions** might require writing additional code to maintain state (for example, using lists or other collections to store results).

### Example:
- A generator function can be written as a simple `def` function using the `yield` keyword, which makes the code concise and easy to understand compared to the alternative iterator pattern involving explicit `__iter__()` and `__next__()` methods.

---

##  **5. Handling Infinite Sequences**

- **Generators can handle infinite sequences** like an infinite series or a stream of data from a real-time source.
- Since they compute values **on demand**, they can generate an **infinite number of items** without running out of memory or requiring an infinite amount of space.

### Example:
- Generating Fibonacci numbers or reading an endless stream of data from a network connection can be done using a generator, without worrying about memory constraints.

---

##  **6. Compatibility with Loops**

- **Generators integrate seamlessly with `for` loops** in Python.
- When iterating over a generator, Python will automatically call `next()` and handle the **`StopIteration`** exception when the generator is exhausted.
- This makes generators highly **readable and easy to use** in loops, just like other iterables (e.g., lists or tuples).

---

### **Summary of Advantages:**

| **Advantage**                   | **Explanation**                                                                                                                                       |
|----------------------------------|-------------------------------------------------------------------------------------------------------------------------------------------------------|
| **Memory Efficiency**            | Generates values on demand, reducing memory usage, as only one value is stored at any time.                                                           |
| **Lazy Evaluation**              | Delays computation until the value is requested, improving performance when dealing with large or unnecessary computations.                            |
| **State Preservation**           | Automatically remembers the state between calls, making it easy to resume execution without complex logic.                                            |
| **Simplified Code**              | Cleaner and simpler syntax, requiring less code compared to regular functions or manual iterators.                                                   |
| **Handling Infinite Sequences** | Supports generating infinite sequences, like the Fibonacci series or continuous data streams, without running out of memory.                         |
| **Loop Integration**             | Easy to use with Python's `for` loops, enhancing code readability and making iteration seamless.                                                      |

 ###  ** Purpose and Usage of the `map()` Function in Python**

The `map()` function in Python is a **built-in function** used to **apply a given function to each item** in an iterable (such as a list, tuple, or set) and return a **new map object** (an iterator) containing the results.

---

##  **Purpose of `map()` Function:**

The primary purpose of `map()` is to **transform** all items in an iterable **without writing an explicit loop**.  
It is commonly used for applying **the same operation** to every element in a sequence.

---

##  **Syntax:**

```python
map(function, iterable)
```

- **`function`**: A function (can be predefined, user-defined, or a lambda) that processes each item.
- **`iterable`**: A sequence like a list, tuple, or set whose items will be passed to the function.

---

##  **Key Characteristics:**

| Feature                   | Description                                                    |
|---------------------------|----------------------------------------------------------------|
| **Functional Programming**| `map()` supports functional programming concepts               |
| **Returns Iterator**      | The result must be converted to a list/tuple for viewing       |
| **Applies Function**      | Each element is passed to the given function                   |

---

##  **How It Works:**

- `map()` takes each item from the iterable.
- Passes it to the given function.
- Collects and returns the modified results as a **map object**.

---

##  **Example (Theoretical):**

> Suppose you want to double all numbers in a list.

Instead of using a `for` loop, you can use `map()`:

```python
numbers = [1, 2, 3, 4]
result = map(double, numbers)
```

- Here, `double` is a function that takes a number and returns it multiplied by 2.
- `map(double, numbers)` applies `double()` to each element of `numbers`.

---

##  **Typical Use Cases:**

1. **Mathematical operations on all items** in a list.

````python
def sum_of_even_numbers(numbers):
    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

# Example usage:
my_list = [10, 15, 20, 25, 30]
result = sum_of_even_numbers(my_list)
print("Sum of even numbers:", result)

#(Written for input: [10, 15, 20, 25, 30] → Even numbers are 10, 20, 30 → Total = 60)


````

#Q4.Write a python  function that checks if a given number is prime or not from 1 to 200.

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage: Check prime numbers from 1 to 200
for number in range(1, 201):
    if is_prime(number):
        print(f"{number} is a prime number.")

````

#Q5.Create an iterator class in python that generates the Fibonacci sequence up to a specified number of terms.

````python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return self.a
        elif self.count == 1:
            self.count += 1
            return self.b
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.b

# Example usage:
fib = FibonacciIterator(10)
for number in fib:
    print(number)
````

#Q6.Write a generator function in python that yields the powers of 2 up to given exponent.

````python
def powers_of_two(max_exponent):
    for exponent in range(max_exponent + 1):
        yield 2 ** exponent

# Example usage:
for value in powers_of_two(5):
    print(value)
````

#Q7.Implement a generator that reads a file in line and yields each  line as a string.

````python
def line_generator(file_path):
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line  # Yield each line as it is
    except FileNotFoundError:
        print("The file was not found.")
        return
    except Exception as e:
        print(f"An error occurred: {e}")
        return

# Example usage:
file_path = 'example.txt'  # Ensure this file exists
for line in line_generator(file_path):
    print(line, end='')  # end='' is used to avoid double newline



````
